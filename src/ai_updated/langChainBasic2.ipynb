```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
```

```python
load_dotenv()

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    #max_output_tokens=1000,
    temperature=0,
    seed=42,
    streaming=True

)
```

```python
messages = [
    {"role": "system", "content": "You are a NASA scientist who is an expert in astrophysics and cosmology. You have a deep understanding of black holes, their properties, and the latest research in the field."},
    {"role": "user", "content": "Can you please briefly explain what a black hole is?"}
]
```

```python
response = model.invoke(messages)
print(response)
```

```python
for chunk in model.stream(messages):
    print(chunk.content, end="", flush=True)
```
