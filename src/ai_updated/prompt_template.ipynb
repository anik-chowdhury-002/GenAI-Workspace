```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
```

```python
load_dotenv()

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash"

)
```

Prompt Template

```python
TEMPLATE = '''
System: {description}

Human: I've recenntly adopted a {pet}. Could you suggest some {pet} names ?

'''

prompt_template = PromptTemplate.from_template(template=TEMPLATE)
```

```python
prompt_template
```

```python
print(prompt_template.template)
```

```python
prompt_value = prompt_template.invoke({'description': "The Chatbot should reluctantly answer questions with sarcastic responses.",
                                       'pet':'dog'})
```

```python
print(prompt_value.text)
```

```python
response = model.invoke(prompt_value)
print(response.content)
```

ChatPrompt Templete

```python
from langchain_core.prompts.chat import (SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate, ChatPromptTemplate)
```

```python
TEMPLATE_S = '{description}'
TEMPLATE_H = '''I've recenntly adopted a {pet}. Could you suggest some {pet} names ?'''
```

```python
message_template_s = SystemMessagePromptTemplate.from_template(template=TEMPLATE_S)
message_template_h = HumanMessagePromptTemplate.from_template(template=TEMPLATE_H)
```

```python
message_template_s
```

```python
message_template_h
```

```python
chat_template = ChatPromptTemplate.from_messages([message_template_s, message_template_h])
```

```python
chat_template
```

```python
chat_value = chat_template.invoke({'description': "The Chatbot should reluctantly answer questions with sarcastic responses.",
                                       'pet':'dog'})
```

```python
chat_value
```

```python
print(chat_value.messages)
```

```python
response = model.invoke(chat_value)
print(response.content)
```

Few-Shot Prompting (FewSortChatMessagePromptTemplate)

```python
from langchain_core.prompts import FewShotChatMessagePromptTemplate
```

```python
TEMPLATE_H = '''I've recenntly adopted a {pet}. Could you suggest some {pet} names ?'''
TEMPLATE_AI = '''{response}'''
```

```python
message_template_h = HumanMessagePromptTemplate.from_template(template=TEMPLATE_H)
message_template_ai = AIMessagePromptTemplate.from_template(template=TEMPLATE_AI)
```

```python
exaample_template = ChatPromptTemplate.from_messages([message_template_h, message_template_ai])
```

```python
examples = [{'pet':'cat',
             'response' : ''' 1.Meow 2.Kitty 3.Cutey 4.Fuchulu 5.Kuchulu '''
            },

            {'pet':'dog',
             'response' : ''' 1.Boss 2.Guddu 3.Loku 4.Tommy 5.Tom '''
            }]
```

```python
few_shot = FewShotChatMessagePromptTemplate(examples=examples, example_prompt=exaample_template)
```

```python
print(few_shot.examples)
```

```python
chat_template = ChatPromptTemplate.from_messages([few_shot, message_template_h])
```

```python
chat_template
```

```python
chat_value = chat_template.invoke({'pet':'rabbit'})
```

```python
chat_value
```

```python
for i in chat_value.messages:
    print(f' {i.type}:{i.content}')
```

```python
response = model.invoke(chat_value)
```

```python
print(response.content)
```
