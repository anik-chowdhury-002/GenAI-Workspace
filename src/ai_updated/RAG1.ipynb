Document Splitter

Document Splitting with CharacterTextSplitter

```python
from langchain_text_splitters.character import CharacterTextSplitter
```

```python
total_length = 0
for i in range(len(pages_copy)):
    len(pages_copy[i].page_content)
    total_length += len(pages_copy[i].page_content)
print(total_length)
```

```python
character_splitter = CharacterTextSplitter(
    separator = " ",
    chunk_size = 500,
    chunk_overlap  = 50)
```

```python
pages_character_split = character_splitter.split_documents(pages_copy)
```

```python
len(pages_character_split[0].page_content),len(pages_character_split[1].page_content),len(pages_character_split[2].page_content),len(pages_character_split[3].page_content),len(pages_character_split[4].page_content),len(pages_character_split[5].page_content),len(pages_character_split[6].page_content),len(pages_character_split[7].page_content),len(pages_character_split[8].page_content),len(pages_character_split[9].page_content)
```

Document Splitting With MarkdownHeaderTextSplitter

```python
from langchain_text_splitters.markdown import MarkdownHeaderTextSplitter
```

```python
md_splitter = MarkdownHeaderTextSplitter(headers_to_split_on = [("#", "Heading 1"),
                                                                ("##", "Heading 2")])
```

```python
pages_md_split = md_splitter.split_text(pages_copy[0].page_content)
```

Embeddings

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python

embedding = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2",
    #dimensions=32
)
```

```python
vector1 = embedding.embed_query(pages_character_split[0].page_content)
vector2 = embedding.embed_query(pages_character_split[1].page_content)
vector3 = embedding.embed_query(pages_character_split[2].page_content)
```

```python
len(vector1),len(vector2),len(vector3)
```

```python
import numpy as np
```

```python
print(np.dot(vector1,vector2))
print(np.dot(vector2,vector3))
print(np.dot(vector1,vector3))
```

```python
np.linalg.norm(vector1),np.linalg.norm(vector2),np.linalg.norm(vector3)
```
