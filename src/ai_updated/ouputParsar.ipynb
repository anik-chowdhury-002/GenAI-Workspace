```python
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
chat = ChatGroq(
    model="allam-2-7b",
    max_tokens=100,
    temperature=0,
    seed=42
)

message_h = HumanMessage(content = "What is ML")
response = chat.invoke([message_h])
print(response.content)
```

```python
str_output_parser = StrOutputParser()
parsed_response = str_output_parser.invoke(response)
parsed_response
```

Comma Separated List Output ParSer

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
message_h = HumanMessage("I have recently adoped a male white cat, can you please suggest some creative name?")
response = chat.invoke([message_h])
print(response.content)

```

```python
list_output_parsar = CommaSeparatedListOutputParser()
parsed_response = list_output_parsar.invoke(response)
print(parsed_response)
```

```python
print(list_output_parsar.get_format_instructions())
```

```python
message_h = HumanMessage(
    '''I have recently adoped a male white cat, can you please suggest some creative name?
    {CommaSeparatedListOutputParser().get_format_instructions()} ''')

print(message_h.content)
```

```python
response = chat.invoke([message_h])
response.content
```

```python
parsed_response = list_output_parsar.invoke(response)
print(parsed_response)
```

Date Time parser

```python
from langchain_classic.output_parsers import DatetimeOutputParser
from langchain_core.prompts import PromptTemplate
```

```python
# need to initialize the parser (optional: provide custon date time format , default format is y-m-d h:m:s)

output_parser = DatetimeOutputParser()

prompt = PromptTemplate(
    template="Answer the users question. \n{format_instructions}\n{question}",
    input_variables=["question"],
    partial_variables= {"format_instructions":output_parser.get_format_instructions()},
)


model = ChatGroq(
    model="allam-2-7b",
    max_tokens=100,
    temperature=0
)


chain = prompt | model | output_parser

response = chain.invoke({"question": " When was 1st iphone released?"})


print(type(response))
print(response)
```
