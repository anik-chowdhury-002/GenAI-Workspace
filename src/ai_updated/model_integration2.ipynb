ধরুন, আপনি অনলাইনে খাবার অর্ডার করতে চান। আপনার কাছে দুটো উপায় আছে:

init_chat_model (Swiggy/Zomato-এর মতো)
সহজ কথা: এটি LangChain-এর একটি "All-in-one" বা সর্বজনীন পদ্ধতি।

কীভাবে কাজ করে: Swiggy-তে যেমন আপনি শুধু লিখে দেন "KFC থেকে বার্গার চাই" বা "Domino's থেকে পিৎজা চাই", ঠিক তেমনি এখানে আপনি শুধু কোডে লিখে দিচ্ছেন "google_genai:gemini-3.5-flash"। LangChain নিজেই বুঝে নেয় গুগলের সাথে কীভাবে যোগাযোগ করতে হবে।

সবচেয়ে বড় সুবিধা: ধরুন কাল আপনার মনে হলো, "না, আমি গুগলের Gemini ব্যবহার করব না, OpenAI-এর ChatGPT ব্যবহার করব।" তখন আপনাকে পুরো কোড পাল্টাতে হবে না। শুধু নামটা বদলে "openai:gpt-4o" করে দিলেই কাজ হয়ে যাবে। আপনার অ্যাপ্লিকেশনে কোনো ভাঙন ধরবে না।

```python
model = init_chat_model('google_genai:gemini-3.5-flash')
response = model.invoke("Why do parrots talk?")
response.content
```

ChatGoogleGenerativeAI (Domino's-এর নিজস্ব অ্যাপের মতো)
সহজ কথা: এটি হলো গুগলের মডেলকে ডাকার সরাসরি (Direct) পদ্ধতি।

কীভাবে কাজ করে: এখানে আপনি সরাসরি গুগলের জন্য তৈরি করা নির্দিষ্ট ক্লাস (ChatGoogleGenerativeAI) ইমপোর্ট করে কাজ করছেন। এটা ঠিক Domino's-এর নিজস্ব অ্যাপ ডাউনলোড করে সেখান থেকে অর্ডার করার মতো।

অসুবিধা: কাল যদি আপনি ChatGPT ব্যবহার করতে চান, তবে এই কোড আর কাজ করবে না। আপনাকে আবার OpenAI-এর জন্য আলাদা ক্লাস (ChatOpenAI) ইমপোর্ট করে নতুন করে কোড লিখতে হবে।

সুবিধা: যেহেতু এটি সরাসরি গুগলের নিজস্ব ক্লাস, তাই গুগলের মডেলের যদি কোনো একদম স্পেশাল ফিচার থাকে (যা অন্য মডেলে নেই), সেটা এই পদ্ধতিতে খুব নিখুঁতভাবে কন্ট্রোল করা যায়।

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model='gemini-3.5-flash')
response = model.invoke("Why do parrots talk?")
response
```

আউটপুটের সামান্য পার্থক্য কেন?
প্রথম ক্ষেত্রে আপনি সরাসরি response.content প্রিন্ট করেছেন, যা কাঁচা তালিকা/টেক্সট আউটপুট দিয়েছে।

দ্বিতীয় ক্ষেত্রে আপনি পুরো response অবজেক্টটি প্রিন্ট করেছেন, তাই এটি একটি নির্দিষ্ট AIMessage(...) অবজেক্ট হিসেবে দেখাচ্ছে। (response.content লিখলে সেখানেও টেক্সটটিই পেতেন)।
